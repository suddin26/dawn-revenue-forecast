# Load DAWN revenue history
Read the combined CSV into df. Requires pandas. This notebook loads all rows without filtering or recoding.


```python
from pathlib import Path
import pandas as pd

# Locate the CSV from the project root or the notebook folder.
filename = 'dawn_revenue_accounts_combined_history.csv'
data_path = next(
    (folder / 'data' / filename
     for folder in [Path.cwd(), *Path.cwd().parents]
     if (folder / 'data' / filename).is_file()),
    None,
)
if data_path is None:
    raise FileNotFoundError(f'Cannot find data/{filename}. Start Jupyter inside the project folder.')

# Preserve account identifiers as text, including any leading zeros.
identifier_columns = ['Doc Number', 'level_1_id', 'level_2_id', 'GL Number', 'BA Number']
df = pd.read_csv(
    data_path,
    dtype={column: 'string' for column in identifier_columns},
    low_memory=False,
)
df['Date'] = pd.to_datetime(df['Date'], format='%m/%d/%Y', errors='raise')

print(f'Reading: {data_path}')
print(f'Loaded {len(df):,} rows and {len(df.columns)} columns.')
df.head(10)
```
